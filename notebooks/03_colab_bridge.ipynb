## Pregunta
¿Las transacciones con fraude tienen montos más altos que las transacciones normales?

## Datos
Los datos son **sintéticos** (inventados con código). Se crean dentro de este mismo notebook con NumPy y una semilla fija (`42`), así que siempre salen iguales. No se usan datos reales ni personales, y no se carga ningún archivo ni Drive.

```python
# Configuración: versión de Python y librerías necesarias
# Colab ya trae instaladas estas librerías, así que solo se importan.
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print("Python:", sys.version.split()[0])
print("numpy:", np.__version__)
print("pandas:", pd.__version__)
print("matplotlib:", matplotlib.__version__)
```

```python
# Datos sintéticos: 200 transacciones inventadas
# La semilla fija (42) hace que siempre salgan los mismos datos.
rng = np.random.default_rng(42)

n = 200
es_fraude = rng.random(n) < 0.10  # aproximadamente 10% de las transacciones son fraude

# Las normales giran alrededor de $500 y las de fraude alrededor de $1,500
monto = np.where(
    es_fraude,
    rng.normal(1500, 400, n),
    rng.normal(500, 150, n),
).clip(10).round(2)  # sin montos negativos, con 2 decimales

datos = pd.DataFrame({
    "id_transaccion": range(1, n + 1),
    "monto": monto,
    "tipo": np.where(es_fraude, "Fraude", "Normal"),
})

print("Filas y columnas:", datos.shape)
datos.head()
```

```python
# Análisis: comparar el monto promedio entre fraude y normales
resumen = (
    datos.groupby("tipo")["monto"]
    .agg(transacciones="count", monto_promedio="mean")
    .round(2)
)
print(resumen)

# Gráfica de barras con el monto promedio de cada tipo
resumen["monto_promedio"].plot(kind="bar", color=["tab:red", "tab:blue"])
plt.title("Monto promedio por tipo de transacción (datos sintéticos)")
plt.xlabel("Tipo de transacción")
plt.ylabel("Monto promedio ($)")
plt.xticks(rotation=0)
plt.show()
```

## Conclusión
Sí. En estos datos sintéticos, las transacciones con fraude tienen un monto promedio de aproximadamente **$1,400**, contra unos **$495** de las normales, o sea casi 3 veces más. Solo 18 de las 200 transacciones (9%) son fraude.
